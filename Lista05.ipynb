CALSSE DOUBLYLINKEDLIST GERAL

```python
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None
        self.anterior = None


class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def contem(self, valor):
        atual = self.head
        while atual is not None:
            if atual.valor == valor:
                return True
            atual = atual.proximo
        return False

    def obter_intervalo(self):
        if self.head is None:
            return None, None

        minimo = self.head.valor
        maximo = self.head.valor
        atual = self.head.proximo

        while atual is not None:
            if atual.valor < minimo:
                minimo = atual.valor
            if atual.valor > maximo:
                maximo = atual.valor
            atual = atual.proximo

        return minimo, maximo

    def ins_final(self, valor):
        novo_no = Node(valor)


        if self.head is None:
            self.head = novo_no
            self.tail = novo_no
            return


        novo_no.anterior = self.tail
        self.tail.proximo = novo_no
        self.tail = novo_no

    def exibir_frente(self):
        atual = self.head
        print("Normal: ", end="")
        while atual is not None:
            print(f"[{atual.valor}]", end=" <-> ")
            atual = atual.proximo
        print("None")

    def exibir_tras(self):
        atual = self.tail
        print("Invertida: ", end="")
        while atual is not None:
            print(f"[{atual.valor}]", end=" <-> ")
            atual = atual.anterior
        print("None")

    def elemento_central_simples(self):

        if self.head is None:
            print("A lista está vazia.")
            return None


        tamanho = 0
        atual = self.head
        while atual is not None:
            tamanho += 1
            atual = atual.proximo

        indice_meio = (tamanho - 1) // 2

        atual = self.head
        for i in range(indice_meio):
            atual = atual.proximo

        if tamanho == 0:
          print("Lista sem elemento central")
          return

        else:

         print(f"Elemento central: {atual.valor}")
         return atual.valor

    def inserir_ordenado(self, valor):
        novo_no = Node(valor)

        if self.head is None:
            self.head = novo_no
            self.tail = novo_no
            return

        if valor <= self.head.valor:
            novo_no.proximo = self.head
            self.head.anterior = novo_no
            self.head = novo_no
            return

        if valor >= self.tail.valor:
            novo_no.anterior = self.tail
            self.tail.proximo = novo_no
            self.tail = novo_no
            return

        atual = self.head
        while atual is not None and atual.valor < valor:
            atual = atual.proximo

        novo_no.proximo = atual
        novo_no.anterior = atual.anterior

        atual.anterior.proximo = novo_no
        atual.anterior = novo_no

    def atualizar(self, y):

        if self.head is None:
            print("A lista está vazia! Não há nós para atualizar.")
            return

        nos = []
        atual = self.head
        while atual is not None:
            nos.append(atual)
            atual = atual.proximo

        no_sorteado = random.choice(nos)
        x = no_sorteado.valor

        print(f"\nO nó sorteado aleatoriamente tem o valor: x = {x}")

        if no_sorteado == self.head:
            self.head = no_sorteado.proximo
            if self.head is not None:
                self.head.anterior = None

        elif no_sorteado == self.tail:
            self.tail = no_sorteado.anterior
            if self.tail is not None:
                self.tail.proximo = None

        else:
            no_sorteado.anterior.proximo = no_sorteado.proximo
            no_sorteado.proximo.anterior = no_sorteado.anterior

        self.inserir_ordenado(y)
        print(f"Sucesso! O valor {x} foi atualizado para {y} e reposicionado.")

    def trocar_nos(self, no_a, no_b):

        ante_a = no_a.anterior
        prox_b = no_b.proximo

        if ante_a is not None:
            ante_a.proximo = no_b
        else:
            self.head = no_b

        if prox_b is not None:
            prox_b.anterior = no_a
        else:
            self.tail = no_a

        no_b.anterior = ante_a
        no_b.proximo = no_a

        no_a.anterior = no_b
        no_a.proximo = prox_b

    def varredura(self):
        if self.head is None or self.head.proximo is None:
            return

        atual = self.head

        while atual is not None and atual.proximo is not None:
            proximo_no = atual.proximo

            if atual.valor > proximo_no.valor:
                self.trocar_nos(atual, proximo_no)
            else:
                atual = atual.proximo

    def particao(self, k):
        if self.head is None or self.head == self.tail:
            return

        q = self.head
        r = self.tail

        while q != r and q.anterior != r:

            while q != r and q.valor <= k:
                q = q.proximo

            while q != r and r.valor > k:
                r = r.anterior

            if q != r and q.anterior != r and q.valor > k and r.valor <= k:

                q.valor, r.valor = r.valor, q.valor

                q = q.proximo
                r = r.anterior

```

```python
def transformar_em_lista_de_listas(lista_original, k):
    if lista_original.head is None or k <= 0:
        return []

    N = 0
    atual = lista_original.head
    while atual is not None:
        N += 1
        atual = atual.proximo

    tam_sublista = N // k
    resto = N % k

    vetor_de_sublistas = []
    atual = lista_original.head

    for i in range(k):
        sublista = DoublyLinkedList()
        sublista.head = atual

        tamanho_atual = tam_sublista + (1 if i < resto else 0)

        for _ in range(1, tamanho_atual):
            if atual is not None:
                atual = atual.proximo

        if atual is not None:
            sublista.tail = atual
            proxima_sublista = atual.proximo

            atual.proximo = None
            if proxima_sublista is not None:
                proxima_sublista.anterior = None

            atual = proxima_sublista

        vetor_de_sublistas.append(sublista)

    return vetor_de_sublistas
```

```python
def busca_lista_de_listas(L, x):
    sub_alvo = -1

    for i, sublista in enumerate(L):
        if sublista.head is not None:
            if x <= sublista.tail.valor:
                sub_alvo = i
                break

    if sub_alvo == -1:
        return None

    atual = L[sub_alvo].head
    while atual is not None:
        if atual.valor == x:
            return atual
        if atual.valor > x:
            return None
        atual = atual.proximo

    return None

def insercao_lista_de_listas(L, x):
    sub_alvo = len(L) - 1

    for i, sublista in enumerate(L):
        if sublista.head is not None and x <= sublista.tail.valor:
            sub_alvo = i
            break

    sublista_destino = L[sub_alvo]
    novo_no = Node(x)

    if sublista_destino.head is None:
        sublista_destino.head = novo_no
        sublista_destino.tail = novo_no
        return

    atual = sublista_destino.head
    while atual is not None and atual.valor < x:
        atual = atual.proximo

    if atual == sublista_destino.head:
        novo_no.proximo = sublista_destino.head
        sublista_destino.head.anterior = novo_no
        sublista_destino.head = novo_no
    elif atual is None:
        novo_no.anterior = sublista_destino.tail
        sublista_destino.tail.proximo = novo_no
        sublista_destino.tail = novo_no
    else:
        novo_no.proximo = atual
        novo_no.anterior = atual.anterior
        atual.anterior.proximo = novo_no
        atual.anterior = novo_no

def remocao_lista_de_listas(L, x):
    no_alvo = busca_lista_de_listas(L, x)
    if no_alvo is None:
        return

    for sublista in L:
        if (
            sublista.head is not None
            and sublista.head.valor <= x <= sublista.tail.valor
        ):
            if no_alvo.anterior is not None:
                no_alvo.anterior.proximo = no_alvo.proximo
            else:
                sublista.head = no_alvo.proximo

            if no_alvo.proximo is not None:
                no_alvo.proximo.anterior = no_alvo.anterior
            else:
                sublista.tail = no_alvo.anterior
            break
```

ESPARSOS

```python
class NoEsparso:
    def __init__(self, valor, posicao):
        self.valor = valor
        self.posicao = posicao
        self.proximo = None
        self.anterior = None


class ListaVetorEsparso:
    def __init__(self):
        self.head = None
        self.tail = None

    def inserir_no_final(self, valor, posicao):
        novo_no = NoEsparso(valor, posicao)

        if self.head is None:
            self.head = novo_no
            self.tail = novo_no
        else:
            novo_no.anterior = self.tail
            self.tail.proximo = novo_no
            self.tail = novo_no

    def construir_de_vetor(self, vetor):
        for i in range(len(vetor)):
            if vetor[i] != 0:
                posicao_original = i + 1
                self.inserir_no_final(vetor[i], posicao_original)

    def exibirEspr(self):
        atual = self.head
        print("Lista Esparsa: ", end="")
        while atual is not None:
            print(f"[{atual.valor} | pos:{atual.posicao}]", end=" <-> ")
            atual = atual.proximo
        print("None")

    def busca_por_indice(self, k):
     atual = self.head
     while atual is not None and atual.posicao <= k:
        if atual.posicao == k:
            return atual.valor
        atual = atual.proximo
     return 0


    def busca_por_valor(self, x, tamanho_vetor):
        if x == 0:
            pos_esperada = 1
            atual = self.head
            while atual is not None:
                if atual.posicao > pos_esperada:
                    return pos_esperada
                pos_esperada = atual.posicao + 1
                atual = atual.proximo
            return pos_esperada if pos_esperada <= tamanho_vetor else -1

        atual = self.head
        while atual is not None:
            if atual.valor == x:
                return atual.posicao
            atual = atual.proximo
        return -1

    def atualiz(self, x, k):
        atual = self.head

        while atual is not None and atual.posicao < k:
            atual = atual.proximo

        if atual is not None and atual.posicao == k:
            if x != 0:
                atual.valor = x
            else:
                if atual.anterior is not None:
                    atual.anterior.proximo = atual.proximo
                else:
                    self.head = atual.proximo

                if atual.proximo is not None:
                    atual.proximo.anterior = atual.anterior
                else:
                    self.tail = atual.anterior

        elif x != 0:
            novo_no = NoEsparso(x, k)

            if atual == self.head:
                novo_no.proximo = self.head
                if self.head is not None:
                    self.head.anterior = novo_no
                self.head = novo_no
                if self.tail is None:
                    self.tail = novo_no

            elif atual is None:
                novo_no.anterior = self.tail
                if self.tail is not None:
                    self.tail.proximo = novo_no
                self.tail = novo_no

            else:
                novo_no.proximo = atual
                novo_no.anterior = atual.anterior
                atual.anterior.proximo = novo_no
                atual.anterior = novo_no
```
