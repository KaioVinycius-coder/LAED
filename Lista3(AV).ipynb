```python
from random import randint
def creat_vector(t, limi,lims):
  vector = [0] * t
  for i in range(t):
    vector[i] = random.randint(limi , lims)

  return vector
```

```python
def ord(vector):
  for i in range(len(vector)):
    for j in range(i+1, len(vector)):
      if(vector[i] > vector[j]):
        aux = vector[j]
        vector[j] = vector[i]
        vector[i] = aux

  return vector
```

```python
from random import randint
def creat_mat(t , c, limi,lims):
  mat = []
  for i in range(t):
    lin = []
    for j in range(c):
     num = randint(limi , lims)
     lin.append(num)

    mat.append(lin)

  return mat
```
