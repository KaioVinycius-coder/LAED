CLASSE LINKEDLIST GERAL

```python
class Node:

    def __init__(self, valor):
        self.valor = valor
        self.proximo = None


class LinkedList:

    def __init__(self):
        self.head = None

    def inserir_no_final(self, valor):

        novo_no = Node(valor)

        if self.head is None:
            self.head = novo_no
            return

        atual = self.head
        while atual.proximo is not None:
            atual = atual.proximo

        atual.proximo = novo_no

    def inserir_ordenado(self, valor):

        novo_no = Node(valor)

        if self.head is None or valor < self.head.valor:
            novo_no.proximo = self.head
            self.head = novo_no
            return

        atual = self.head

        while atual.proximo is not None and atual.proximo.valor < valor:
            atual = atual.proximo

        novo_no.proximo = atual.proximo
        atual.proximo = novo_no

    def exibir(self):
        """Imprime a lista de forma legível."""
        atual = self.head
        while atual is not None:
            print(f"[{atual.valor}]", end=" -> ")
            atual = atual.proximo
        print("None")

    def mai_fin(self):

        if self.head is None or self.head.proximo is None:
            return

        mai_va = self.head.valor
        alvo = self.head
        ant_alvo = None

        atual = self.head
        while atual.proximo is not None:
            if atual.proximo.valor > mai_va:
                mai_va = atual.proximo.valor
                alvo = atual.proximo
                ant_alvo = atual
            atual = atual.proximo

        if alvo.proximo is None:
            return

        if alvo == self.head:
            self.head = self.head.proximo
        else:
            ant_alvo.proximo = alvo.proximo

        ultimo = self.head
        while ultimo.proximo is not None:
            ultimo = ultimo.proximo

        ultimo.proximo = alvo
        alvo.proximo = None

    def sep_par_imp(self):

        lista_pares = LinkedList()
        lista_impares = LinkedList()

        atual = self.head

        while atual is not None:
            if atual.valor % 2 == 0:
                lista_pares.inserir_no_final(atual.valor)
            else:
                lista_impares.inserir_no_final(atual.valor)

            atual = atual.proximo

        return lista_pares, lista_impares

    def inverter(self):
        ant = None
        atual = self.head

        while atual is not None:

            proximo_no = atual.proximo

            atual.proximo = ant

            ant = atual

            atual = proximo_no

        self.head = ant

    def dup_imp(self):

        atual = self.head

        while atual is not None:

            if atual.valor % 2 != 0:

                new = Node(atual.valor)

                new.proximo = atual.proximo

                atual.proximo = new

                atual = new.proximo
            else:

                atual = atual.proximo

    def intercalar(self, outra_lista):

        lista_intercalada = LinkedList()

        p1 = self.head
        p2 = outra_lista.head

        while p1 is not None and p2 is not None:
            if p1.valor <= p2.valor:
                lista_intercalada.inserir_no_final(p1.valor)
                p1 = p1.proximo
            else:
                lista_intercalada.inserir_no_final(p2.valor)
                p2 = p2.proximo

        while p1 is not None:
            lista_intercalada.inserir_no_final(p1.valor)
            p1 = p1.proximo

        while p2 is not None:
            lista_intercalada.inserir_no_final(p2.valor)
            p2 = p2.proximo

        return lista_intercalada

    def particao(self, k):

        lista_menores = LinkedList()
        lista_maiores = LinkedList()

        atual = self.head

        while atual is not None:
            if atual.valor <= k:
                lista_menores.inserir_no_final(atual.valor)
            else:
                lista_maiores.inserir_no_final(atual.valor)
            atual = atual.proximo

        if lista_menores.head is None:
            self.head = lista_maiores.head
            return

        fim_menores = lista_menores.head
        while fim_menores.proximo is not None:
            fim_menores = fim_menores.proximo

        fim_menores.proximo = lista_maiores.head

        self.head = lista_menores.head

    def remover_todas_copias(self, k):
        while self.head is not None and self.head.valor == k:
            self.head = self.head.proximo

        atual = self.head
        while atual is not None and atual.proximo is not None:
            if atual.proximo.valor == k:

                atual.proximo = atual.proximo.proximo
            else:
                atual = atual.proximo

    def elementos_mais_frequentes(self):

        if self.head is None:
            return [],

        contagem = {}
        atual = self.head

        while atual is not None:
            valor = atual.valor
            contagem[valor] = contagem.get(valor, 0) + 1
            atual = atual.proximo

        max_ocorrencias = max(contagem.values())

        if max_ocorrencias == 1:
            return [], 1

        modas = [valor for valor, qte in contagem.items() if qte == max_ocorrencias]

        if len(modas) == len(contagem):
            return [], max_ocorrencias

        return modas, max_ocorrencias

    def tem_elemento_repetido(self):

        atual = self.head

        while atual is not None:
            marcador = atual.proximo

            while marcador is not None:
                if marcador.valor == atual.valor:
                    return True
                marcador = marcador.proximo

            atual = atual.proximo

        return False

    def intersecao(self, outra_lista):

        lista_resultado = LinkedList()

        atual_a = self.head

        while atual_a is not None:
            valor_procurado = atual_a.valor

            esta_na_outra = False
            atual_b = outra_lista.head
            while atual_b is not None:
                if atual_b.valor == valor_procurado:
                    esta_na_outra = True
                    break
                atual_b = atual_b.proximo

            if esta_na_outra:
                ja_adicionado = False
                atual_res = lista_resultado.head
                while atual_res is not None:
                    if atual_res.valor == valor_procurado:
                        ja_adicionado = True
                        break
                    atual_res = atual_res.proximo

                if not ja_adicionado:
                    lista_resultado.inserir_no_final(valor_procurado)

            atual_a = atual_a.proximo

        return lista_resultado
```

INSERIR MAIOR NO FINAL

```python
import random

list = LinkedList()

quant = int(input("Quantos nós terá a lista: "))

for i in range(quant):
    va = random.randint(1, 100)
    list.inserir_no_final(va)

print("\nLista Criada: ")
list.exibir()

list.mai_fin()

print("\nLista Apos mover para o final: ")
list.exibir()


```
